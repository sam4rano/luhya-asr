## Adjustments

- **OOM on T4?** Edit `config_files/ASR_train_config_luhya.yaml`: set `batch_size: 1`, `gradient_accumulation_steps: 32`
- **Test quickly?** Set `max_data_hours: 1.0` and `max_steps: 200`
- **Use a different model?** Change `pretrained_model` to `facebook/mms-300m` and `learning_rate` to `0.0005`
